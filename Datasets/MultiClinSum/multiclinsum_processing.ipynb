```python
import pandas as pd
import os

folder_path = r"Datasets\MultiClinSum\raw"
```

```python
leaf_files = []

for current_dir, subdirs, files in os.walk(folder_path):
    if not subdirs:
        for f in files:
            full_path = os.path.join(current_dir, f)
            file_name = full_path.split("\\")[-1]
            print(file_name)
            info_name = file_name.split(".")[0].split("_")
            file_ext = file_name.split(".")[1]
            with open(full_path, "r", encoding="utf-8") as f:
                content = f.read()
                if len(info_name) == 4:
                    result_dict = {"source": info_name[0], "filetype": file_ext, "id": int(info_name[3]), "language": info_name[2], "quality": info_name[1], "isSummary": False, "content": content}
                elif len(info_name) == 5:
                    result_dict = {"source": info_name[0], "filetype": file_ext, "id": int(info_name[3]), "language": info_name[2], "quality": info_name[1], "isSummary": True, "content": content}
                else:
                    result_dict = {"id": int(info_name[3])}
                leaf_files.append(result_dict)
                
df = pd.DataFrame(leaf_files)
```

```python
import pandas as pd

df = pd.read_csv(r"..\Datasets\MultiClinSum\multiclinsum_dataset.csv")
```

```python
df
```

```python
df["complete_id"] = (
    df["id"].astype(str) + "_" +
    df["language"] + "_" +
    df["quality"]
)
df
```

```python
summary_df = df[df["isSummary"] == True]
uns_df = df[df["isSummary"] == False]
```

```python
summary_df
```

```python
uns_df
```

```python
summary_map = summary_df.set_index("complete_id")["content"]

uns_df["summary"] = uns_df["complete_id"].map(summary_map)
```

```python
uns_df
```

```python
uns_df = uns_df[["complete_id", "id", "language", "quality", "source", "filetype", "content", "summary"]]
uns_df = uns_df.sort_values(by=["language", "id", "quality"])
uns_df = uns_df.reset_index(drop=True)
uns_df
```

```python
print(uns_df[uns_df["filetype"] != "txt"].count())
uns_df = uns_df.drop(columns="filetype")
uns_df
```

```python
uns_df["unique_id"] = range(uns_df.shape[0])
uns_df = uns_df[["complete_id", "id", "unique_id", "language", "quality", "source", "content", "summary"]]
uns_df
```

```python
uns_df.to_csv(r"..\Datasets\MultiClinSum\multiclinsum_dataset.csv", index=True)
```
